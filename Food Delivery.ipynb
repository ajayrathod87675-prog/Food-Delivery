## Clean Data

```python
df["Customer_Complaint"]=df["Customer_Complaint"].fillna("No")
df
```

```python
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]
df
```

## EDA / Data Analysis

```python
# Restaurant_Name by orders
df["Restaurant_Name"].value_counts()
```

```python
# City wise total orders
df["City"].value_counts()
```

```python
# City wise average sales
df.groupby("City")["Total_Sales"].mean()
```

```python
# payment_method  average total sales 
df.groupby("Payment_Method")["Total_Sales"].mean()
```

```python
# Restaurant_Name by total sales 
df.groupby("Restaurant_Name")["Total_Sales"].sum()
```

```python
#Sort by Sales (highest to low
df.sort_values(by="Total_Sales", ascending=False)
```

```python
# food item wise total sales 
df.groupby("Food_Item")["Total_Sales"].sum()
```

## Visualization

```python
df.groupby("City")["Total_Sales"].sum().plot(kind="bar")

plt.show()
```

```python
# City by total sales using bar chart
city_sales = df.groupby("City")["Total_Sales"].sum()

city_sales.plot(kind="bar")
plt.title("City by Total Sales")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# food item by total sales
Food_Item = df.groupby("Food_Item")["Total_Sales"].sum()

Food_Item = Food_Item.sort_values(ascending=False).head(10)

Food_Item.plot(kind="bar")
plt.title("Top 10 Food Items by Total Sales")
plt.xlabel("Food Item")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
